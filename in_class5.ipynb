```python
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer
from nltk import pos_tag
from nltk.corpus import wordnet

# Ensure NLTK components are available
import nltk

required = ["punkt", "punkt_tab", "stopwords", "wordnet", "averaged_perceptron_tagger_eng"]
for r in required:
    nltk.download(r)

```

```python
# Load the text file
with open("science.txt", "r") as file:
    raw_text = file.read()

print("RAW TEXT:\n")
print(raw_text)

# Split into lines
lines = raw_text.split("\n")

# Create DataFrame
df = pd.DataFrame(lines, columns=["original"])


# Map POS tags for lemmatization
def get_wordnet_pos(tag):
    if tag.startswith("J"):
        return wordnet.ADJ
    elif tag.startswith("V"):
        return wordnet.VERB
    elif tag.startswith("N"):
        return wordnet.NOUN
    elif tag.startswith("R"):
        return wordnet.ADV
    else:
        return wordnet.NOUN

# Preprocess function
def preprocess(text):
    # Tokenize
    tokens = word_tokenize(text)

    # Lowercase + remove punctuation
    tokens = [t.lower() for t in tokens if t.isalpha()]

    # Remove stopwords
    stop_words = set(stopwords.words("english"))
    tokens = [t for t in tokens if t not in stop_words]

    # POS tagging
    tagged = pos_tag(tokens)

    # Lemmatize
    lemmatizer = WordNetLemmatizer()
    lemmatized = [
        lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tagged
    ]

    return " ".join(lemmatized)
```

```python
# Apply preprocess
df["cleaned"] = df["original"].apply(preprocess)

# Bag of words
vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(df["cleaned"])

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print("\nBAG OF WORDS MATRIX:\n")
display(bow_df)
```

How did preprocessing change your text?
Preprocessing removed all the connective words and then tokenized them, then lemmatized them.  In more human words,
it cut all the words down to their root form and separated them into their own word form.
